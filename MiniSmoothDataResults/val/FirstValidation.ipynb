```python
# Cell 1 - Load stats
import json
import numpy as np
import matplotlib.pyplot as plt

STATS_FILE = '3a_validation_r50.json'  # adjust path if needed
with open(STATS_FILE) as f:
    stats = json.load(f)

layers          = list(stats.keys())
smoothed_layers   = [l for l in layers if stats[l]['smoothed']]
unsmoothed_layers = [l for l in layers if not stats[l]['smoothed']]

print(f'Loaded stats for {len(layers)} layers:')
print(f'  Smoothed:   {len(smoothed_layers)}')
print(f'  Unsmoothed: {len(unsmoothed_layers)}')
print()
print('Smoothed:')
for l in smoothed_layers:
    print(f'  {l}')
print('Unsmoothed:')
for l in unsmoothed_layers:
    print(f'  {l}')
```

```python
# Cell 2 - Summary table: activations (smoothed only)
print('SMOOTHED LAYERS — ACTIVATION SUMMARY')
print(f'{"Layer":<35} {"Std Before":>12} {"Std Theory":>12} {"Std Actual":>12} {"Improvement":>13} {"Flatter?":>10}')
print('-' * 100)
for name in smoothed_layers:
    d      = stats[name]
    before = np.array(d['act_before'])
    theory = np.array(d['act_after_theory'])
    actual = np.array(d['act_after_actual']) if d['act_after_actual'] is not None else None

    std_before  = before.std()
    std_theory  = theory.std()
    std_actual  = actual.std() if actual is not None else float('nan')
    improvement = (std_before - std_actual) / std_before * 100 if actual is not None else float('nan')
    flatter     = ('YES' if std_actual < std_before else 'NO') if actual is not None else 'N/A'

    print(f'{name:<35} {std_before:>12.4f} {std_theory:>12.4f} {std_actual:>12.4f} {improvement:>12.1f}% {flatter:>10}')
```

```python
# Cell 3 - Summary table: activations (unsmoothed — actual should == before)
print('UNSMOOTHED LAYERS — ACTIVATION SUMMARY (actual should match before)')
print(f'{"Layer":<35} {"Std Before":>12} {"Std Actual":>12} {"Diff (should be ~0)":>22}')
print('-' * 85)
for name in unsmoothed_layers:
    d      = stats[name]
    before = np.array(d['act_before'])
    actual = np.array(d['act_after_actual']) if d['act_after_actual'] is not None else None

    std_before = before.std()
    std_actual = actual.std() if actual is not None else float('nan')
    diff       = abs(std_before - std_actual) if actual is not None else float('nan')

    print(f'{name:<35} {std_before:>12.4f} {std_actual:>12.4f} {diff:>22.6f}')
```

```python
# Cell 4 - Summary table: weights (all layers)
print('ALL LAYERS — WEIGHT SUMMARY')
print(f'{"Layer":<35} {"Smoothed":>10} {"Std Before":>12} {"Std Theory":>12} {"Std Actual":>12}')
print('-' * 85)
for name in layers:
    d        = stats[name]
    w_before = np.array(d['weight_before'])
    w_theory = np.array(d['weight_after_theory'])
    w_actual = np.array(d['weight_after_actual'])
    smoothed = 'YES' if d['smoothed'] else 'NO'
    print(f'{name:<35} {smoothed:>10} {w_before.std():>12.4f} {w_theory.std():>12.4f} {w_actual.std():>12.4f}')
```

```python
# Cell 5 - Per-channel plots for SMOOTHED layers
# Layout per layer: 2 rows x 4 cols
#   Row A (activations): [Before vs Theory] [Before vs Actual] [All 3] [Theory - Actual diff]
#   Row B (weights):     [Before vs Theory] [Before vs Actual] [All 3] [Theory - Actual diff]
def plot_per_channel(layer_list, title_prefix, filename):
    n      = len(layer_list)
    if n == 0:
        print(f'No layers to plot for: {title_prefix}')
        return

    n_rows = n * 2
    n_cols = 4
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(24, 4 * n_rows))
    fig.suptitle(f'SmoothQuant: {title_prefix} — Per-Channel Stats', fontsize=14, y=1.001)

    if n == 1:
        axes = axes.reshape(1, 2, n_cols)
    else:
        axes = axes.reshape(n, 2, n_cols)

    for i, name in enumerate(layer_list):
        d = stats[name]

        act_before = np.array(d['act_before'])
        act_theory = np.array(d['act_after_theory'])
        act_actual = np.array(d['act_after_actual']) if d['act_after_actual'] is not None else None
        w_before   = np.array(d['weight_before'])
        w_theory   = np.array(d['weight_after_theory'])
        w_actual   = np.array(d['weight_after_actual'])

        # ── ROW A: Activations ────────────────────────────────────────────
        # A-Col 0: Before vs Theory
        ax = axes[i][0][0]
        ax.plot(act_before, label=f'Before (std={act_before.std():.3f})', alpha=0.8, color='tomato')
        ax.plot(act_theory, label=f'Theory (std={act_theory.std():.3f})', alpha=0.8, color='steelblue')
        ax.set_title(f'{name}\nAct: Before vs Theory')
        ax.set_xlabel('Channel'); ax.set_ylabel('Max |Activation|')
        ax.legend(fontsize=7); ax.grid(True, alpha=0.3)

        # A-Col 1: Before vs Actual
        ax = axes[i][0][1]
        ax.plot(act_before, label=f'Before (std={act_before.std():.3f})', alpha=0.8, color='tomato')
        if act_actual is not None:
            ax.plot(act_actual, label=f'Actual (std={act_actual.std():.3f})', alpha=0.8, color='seagreen')
        ax.set_title(f'{name}\nAct: Before vs Actual')
        ax.set_xlabel('Channel'); ax.set_ylabel('Max |Activation|')
        ax.legend(fontsize=7); ax.grid(True, alpha=0.3)

        # A-Col 2: All 3
        ax = axes[i][0][2]
        ax.plot(act_before, label=f'Before (std={act_before.std():.3f})', alpha=0.8, color='tomato')
        ax.plot(act_theory, label=f'Theory (std={act_theory.std():.3f})', alpha=0.7, color='steelblue', linestyle='--')
        if act_actual is not None:
            ax.plot(act_actual, label=f'Actual (std={act_actual.std():.3f})', alpha=0.8, color='seagreen')
        ax.set_title(f'{name}\nAct: All Three')
        ax.set_xlabel('Channel'); ax.set_ylabel('Max |Activation|')
        ax.legend(fontsize=7); ax.grid(True, alpha=0.3)

        # A-Col 3: Theory - Actual diff
        ax = axes[i][0][3]
        if act_actual is not None:
            diff = act_theory - act_actual
            ax.plot(diff, color='purple', alpha=0.8)
            ax.axhline(0, color='black', linestyle='--', linewidth=0.8)
            ax.text(0.98, 0.98, f'max|diff|={np.abs(diff).max():.2e}',
                    transform=ax.transAxes, ha='right', va='top', fontsize=8,
                    bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))
        else:
            ax.text(0.5, 0.5, 'No actual data', ha='center', va='center', transform=ax.transAxes)
        ax.set_title(f'{name}\nAct: Theory − Actual (flat at 0 = identical)')
        ax.set_xlabel('Channel'); ax.set_ylabel('Difference')
        ax.grid(True, alpha=0.3)

        # ── ROW B: Weights ────────────────────────────────────────────────
        # B-Col 0: Before vs Theory
        ax = axes[i][1][0]
        ax.plot(w_before, label=f'Before (std={w_before.std():.3f})', alpha=0.8, color='tomato')
        ax.plot(w_theory, label=f'Theory (std={w_theory.std():.3f})', alpha=0.8, color='steelblue')
        ax.set_title(f'{name}\nWt: Before vs Theory')
        ax.set_xlabel('Input Channel'); ax.set_ylabel('Max |Weight|')
        ax.legend(fontsize=7); ax.grid(True, alpha=0.3)

        # B-Col 1: Before vs Actual
        ax = axes[i][1][1]
        ax.plot(w_before, label=f'Before (std={w_before.std():.3f})', alpha=0.8, color='tomato')
        ax.plot(w_actual, label=f'Actual (std={w_actual.std():.3f})', alpha=0.8, color='seagreen')
        ax.set_title(f'{name}\nWt: Before vs Actual')
        ax.set_xlabel('Input Channel'); ax.set_ylabel('Max |Weight|')
        ax.legend(fontsize=7); ax.grid(True, alpha=0.3)

        # B-Col 2: All 3
        ax = axes[i][1][2]
        ax.plot(w_before, label=f'Before (std={w_before.std():.3f})', alpha=0.8, color='tomato')
        ax.plot(w_theory, label=f'Theory (std={w_theory.std():.3f})', alpha=0.7, color='steelblue', linestyle='--')
        ax.plot(w_actual, label=f'Actual (std={w_actual.std():.3f})', alpha=0.8, color='seagreen')
        ax.set_title(f'{name}\nWt: All Three')
        ax.set_xlabel('Input Channel'); ax.set_ylabel('Max |Weight|')
        ax.legend(fontsize=7); ax.grid(True, alpha=0.3)

        # B-Col 3: Theory - Actual diff
        ax = axes[i][1][3]
        diff_w = w_theory - w_actual
        ax.plot(diff_w, color='darkorange', alpha=0.8)
        ax.axhline(0, color='black', linestyle='--', linewidth=0.8)
        ax.text(0.98, 0.98, f'max|diff|={np.abs(diff_w).max():.2e}',
                transform=ax.transAxes, ha='right', va='top', fontsize=8,
                bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))
        ax.set_title(f'{name}\nWt: Theory − Actual (flat at 0 = identical)')
        ax.set_xlabel('Input Channel'); ax.set_ylabel('Difference')
        ax.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.savefig(filename, dpi=150, bbox_inches='tight')
    plt.show()


# Plot smoothed layers
plot_per_channel(
    smoothed_layers,
    'SMOOTHED LAYERS (SmoothQuant applied)',
    'smoothquant_smoothed_per_channel.png'
)

```
